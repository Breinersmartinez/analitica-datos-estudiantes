## 3. Para entender qué está pasando · Por qué un gráfico miente

Un gráfico no miente porque los números estén mal. La mentira está en la traducción de número a forma.

### 3.1 Las magnitudes: por qué el eje dice `1e13`

Cuando los números son enormes, matplotlib escribe `1e13` en una esquina. El arreglo es una línea: dividir entre `1e12` y decir "billones de pesos" en la etiqueta del eje.

````python
apropiacion_por_sector = df_limpio.groupby('Nombre Sector')['Apropiación Vigente'].sum()
cinco = apropiacion_por_sector.sort_values(ascending=False).head(5)

fig, ejes = plt.subplots(1, 2, figsize=(14, 4))

ejes[0].barh(range(5), cinco.values, color='#B8C4CE')
ejes[0].set_yticks(range(5))
ejes[0].set_yticklabels(['' for _ in range(5)])
ejes[0].set_title('Pesos crudos: el eje dice 1e13')

ejes[1].barh(range(5), cinco.values / 1e12, color='#1F4E79')
ejes[1].set_yticks(range(5))
ejes[1].set_yticklabels(['' for _ in range(5)])
ejes[1].set_xlabel('Billones de pesos')
ejes[1].set_title('Dividido entre 1e12: el eje se lee')

plt.tight_layout()
plt.show()
````

## 4. La paleta del día y la función que limpia los ejes

Definimos los colores globales y la función auxiliar para limpiar spines y grillas.
Incluimos los tonos neutros y acentos para las estrategias visuales del día.

````python
PALETA = {
    'neutro': '#B8C4CE',
    'enfasis': '#1F4E79',
    'alerta': '#C0392B',
    'referencia': '#7F8C8D',
    'secundario': '#E67E22',
    'gris_neutro': '#799790',
    'verde_esmeralda': '#3EB489',
}

FUENTE = 'Fuente: datos.gov.co - Ejecución del Presupuesto General de la Nación'

plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['figure.dpi'] = 100
plt.rcParams['font.size'] = 11
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.grid'] = False


def limpiar_ejes(ax, eje_grilla='x'):
    """Quita los bordes sobrantes y deja una grilla tenue en un solo eje."""
    for lado in ['top', 'right', 'left']:
        ax.spines[lado].set_visible(False)
    if eje_grilla == 'x':
        ax.xaxis.grid(True, linestyle='--', alpha=0.3)
        ax.yaxis.grid(False)
    else:
        ax.yaxis.grid(True, linestyle='--', alpha=0.3)
        ax.xaxis.grid(False)
    ax.set_axisbelow(True)
    return ax


print('Paleta y estilo definidos.')
````

### 4.1 Los agregados que se van a graficar

Calculamos los resúmenes por Sector y Entidad, así como la referencia de ejecución nacional.

````python
columnas_plata = ['Apropiación Vigente', 'Compromisos', 'Obligaciones', 'Pagos']

sector = df_limpio.groupby('Nombre Sector')[columnas_plata].sum()
sector['% Ejecución'] = sector['Compromisos'] / sector['Apropiación Vigente'] * 100
sector['Apropiación (billones)'] = sector['Apropiación Vigente'] / 1e12
sector['Compromisos (billones)'] = sector['Compromisos'] / 1e12
sector = sector.sort_values('Apropiación Vigente', ascending=False)

entidad = df_limpio.groupby('Nombre Entidad')[['Apropiación Vigente', 'Compromisos']].sum()
entidad['% Ejecución'] = entidad['Compromisos'] / entidad['Apropiación Vigente'] * 100
entidad['Apropiación (billones)'] = entidad['Apropiación Vigente'] / 1e12
entidad = entidad.sort_values('Apropiación Vigente', ascending=False)

ejecucion_nacional = (df_limpio['Compromisos'].sum() /
                      df_limpio['Apropiación Vigente'].sum() * 100)

print(f'Sectores: {len(sector)} | Entidades: {len(entidad)}')
print(f'Ejecución nacional: {ejecucion_nacional:.1f}%')
````

## 7. Caso 3: Honestidad escalar y formato (Eje Y Truncado vs Eje Y en Cero)

**La pregunta:** ¿Cuál es la magnitud real de las diferencias de presupuesto entre las entidades principales?

### Muestra inicial: El gráfico malo (Eje Y Truncado y notación científica `1e12`)

Barras verticales con el eje Y arrancando en 8 (`8e12`) en notación científica sin formato.

````python
# EL GRÁFICO MALO - Eje Y truncado arrancando en 8 con notación científica 1e12
top10_c3 = entidad.head(10).sort_index()

fig, ax = plt.subplots(figsize=(11, 6))

ax.bar(range(len(top10_c3)), top10_c3['Apropiación Vigente'], color='#34495E')
ax.set_ylim(8e12, top10_c3['Apropiación Vigente'].max() * 1.02)

ax.set_xticks(range(len(top10_c3)))
ax.set_xticklabels(top10_c3.index, rotation=90, fontsize=7)
ax.set_ylabel('Apropiación Vigente (pesos)')
ax.set_title('Diferencias de apropiación entre entidades principales', fontsize=16, fontweight='bold')
ax.grid(True, linewidth=1.5, color='gray')

plt.tight_layout()
plt.show()
````

### Diagnóstico & Medición matemática de la mentira

En un gráfico de barras, la altura física de la barra codifica la magnitud absoluta. Si el eje Y arranca en 8 en lugar de 0, la proporción visual de las barras se distorsiona dramáticamente.

Calculamos la **razón real** frente a la **razón aparente** percibida visualmente:

````python
valores_c3 = top10_c3['Apropiación (billones)']
razon_real_c3 = valores_c3.max() / valores_c3.min()
razon_aparente_c3 = (valores_c3.max() - 8) / (valores_c3.min() - 8)

print(f'Razón real entre mayor y menor:     {razon_real_c3:.2f} veces')
print(f'Razón aparente en gráfico truncado: {razon_aparente_c3:.2f} veces')
````

**Pregunta de interpretación 8.** ¿Por qué truncar el eje Y en un gráfico de barras se considera una distorsión grave y no una simple preferencia estilística?

*Tu respuesta:*

<details>
<summary>Comparar con la respuesta esperada</summary>

Porque en un gráfico de barras el área y la altura de la barra representan visualmente la cantidad. Al no iniciar en cero, la relación proporcional de las alturas deja de corresponder a la relación matemática de los datos, engañando al lector sobre la magnitud real de la brecha.

</details>
